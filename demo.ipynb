# HTSIR Quickstart Demo

This notebook demonstrates how to index a document and answer questions with **HTSIR**.

```python
import os

# Set your OpenAI API key (or configure it in your shell environment)
# os.environ["OPENAI_API_KEY"] = "your-openai-api-key"
# Optional: custom OpenAI-compatible base URL
# os.environ["OPENAI_BASE_URL"] = "https://your-endpoint/v1"
```

```python
from htsir import RetrievalAugmentation, RetrievalAugmentationConfig

config = RetrievalAugmentationConfig(
    tb_max_tokens=100,
    use_reranker=True,
    reranker_type="local",
    reranker_model_name="BAAI/bge-reranker-large",
    reranker_top_k=4,
)
ra = RetrievalAugmentation(config=config)
```

```python
with open("demo/sample.txt", "r", encoding="utf-8") as f:
    text = f.read()

ra.add_documents(text, chunk_group_size=2)
```

```python
question = "How did Cinderella reach her happy ending?"
context, answer = ra.answer_question(question=question, top_k=10)
print("Answer:", answer)
```

```python
# Save and reload the constructed tree
os.makedirs("outputs", exist_ok=True)
save_path = "outputs/cinderella.pkl"
ra.save(save_path)

ra_loaded = RetrievalAugmentation(config=config, tree=save_path)
context, answer = ra_loaded.answer_question(question=question, top_k=10)
print("Answer (reloaded):", answer)
```
